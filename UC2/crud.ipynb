```python
class Plataforma():
    nome: str
    fabricante: str


    def __init__(self, nome: str, fabricante: str):
        self.nome=nome
        self.fabricante=fabricante

    def __str__(self):
        return f"Plataforma(nome={self.nome},fabricante={self.fabricante})"

    def __repr__(self):
        return f"Plataforma {self.nome}"   
```

```python
plataformas_para_inserir: list[Plataforma] = [
    Plataforma("Playstation 3", "Sony"),
    Plataforma("Playstation 2", "Sony"),
    Plataforma("Playstation 1", "Sony"),
    Plataforma("XBOX Original", "Microsoft"),
    Plataforma("XBOX 360", "Microsoft"),
    Plataforma("XBOX One", "Microsoft"),
    Plataforma("XBOX Series X", "Microsoft"),  
               
]

plataforma1 = Plataforma("Playstation 3", "Sony")
plataformas_para_inserir.append(plataforma1)
print(plataformas_para_inserir),
```

```python



try:
    cursor.execute(
        "INSERT INTO plataforma (nome, fabricante) VALUES (%s,%s);",
        ("Playstation 4", "Sony")
    )
    conexao.commit
except pg.DatabaseError as e:
    print(e)
    conexao.rollback()




```

```python
for plataforma in plataformas_para_inserir:

    try:
        cursor.execute(
            "INSERT INTO plataforma (nome, fabricante) VALUES (%s,%s);",
            (plataforma.nome, plataforma.fabricante)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()
```
