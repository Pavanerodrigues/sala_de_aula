```python
import psycopg2 as pg  
from datetime import date
```

```python
conexao = pg.connect(
    dbname="pg_personalizados",
    user="postgres",
    password="postgres",
    host="localhost",
    port="5432"

)
```

```python
cursor = conexao.cursor()
print("conectado!!")
```

```python
class Cliente():
    nome: str
    telefone: str
    email: str
    tipo: str


    def __init__(self, nome: str, telefone: str, email: str, tipo: str):
        self.nome=nome
        self.telefone=telefone
        self.email=email
        self.tipo=tipo


    def __str__(self):
        return f"Cliente(nome={self.nome},telefone={self.telefone},email={self.email},tipo={self.tipo})"
    
    def __repr__(self):
        return f"Cliente {self.nome}"    
    
```

```python
clientes: list[Cliente] = [
    Cliente("Solange", "51998454321", "solangeeduc@gmail.com", "professor"),
    Cliente("Vitor", "51996438413", "vitorjog@gmail.com", "aluno"),
    Cliente("Alexandro","5199933254", "alexandroto@gmaail.com", "professor"),
    Cliente("Bruna", "51922284654", "brunachata@gmail.com", "aluno"),
    Cliente("Lilian", "51986314823", "liliandirecao@gmail.com", "diretora"),
    Cliente("Patricia", "5198246534", "patriciasecretaria@gmail.com", "secretaria"),           
]

print(clientes)
```

```python
for cliente in clientes:
    try:
        cursor.execute(
            "INSERT INTO cliente (nome, telefone, email, tipo) VALUES (%s, %s, %s, %s);",
            (cliente.nome, cliente.telefone, cliente.email, cliente.tipo)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()      
```

```python
class Produto():
    nome: str
    tipo: str   
    preco: float
    quantidade_disponivel: int

    def __init__(self, nome:str, tipo: str, preco:float, quantidade_disponivel: int):
        self.nome=nome
        self.tipo=tipo
        self.preco=preco 
        self.quantidade_disponivel=quantidade_disponivel
    
    def __str__(self):
        return f"Produto(nome={self.nome},tipo={self.tipo}, preco={self.preco},quantidade_disponive={self.quantidade_disponivel})"
    
    def __repr__(self):
        return f"Produto {self.tipo, self.preco}"
```

```python
produtos: list[Produto] = [
    Produto("caneca","ceramica", "59.90","200"),
    Produto("caneca","aluminio", "49.90", "100"),
    Produto("caneca","acrilico","29.90", "50"),
    Produto("camiseta", "algodao","49.90", "200"),
    Produto("camiseta","poliester", "59.90", "200"),
]
print(produtos)
```

```python
for produto in produtos:
    try:
        cursor.execute(
            "INSERT INTO produto(nome, tipo, preco, quantidade_disponiovel) VALUES (%s, %s, %s, %s);",
            (produto.nome, produto.tipo, produto.preco, produto.quantidade_disponivel)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()
```

```python
class Encomenda():
    cliente: int
    data_pedido: date
    data_pagamento: date
    data_retirada: date
    descricao: str

    def __init__(self, cliente:int, data_pedido:date, data_pagamento:date, data_retirada:date, descricao:str):
        self.cliente=cliente
        self.data_pedido=data_pedido
        self.data_pagamento=data_pagamento
        self.data_retirada=data_retirada
        self.descricao=descricao

    def __str__(self):
        return f"Encomenda(cliente={self.cliente}, data_pedido={self.data_pedido}, data_pegamento={self.data_pagamento}, data_retirada={self.data_retirada}, descricao={self.descricao})"

    def __repr__(self):
        return f"Encomenda{self.cliente}"    
```

```python
encomenda: list[Encomenda] = [
    Encomenda(1, date(26,3,22), date(26,3,30), date(26,3,30), "estampa Formandos 2026"),
    Encomenda(2, date(26,5,12), date(26,6,2), date(26,6,2),"estampa Formandos 2026"),
    Encomenda(3, date(26,9,25), date(26,10,3), date(26,10,3),"estampa aniersário Arthur")
]
print(encomenda)
```

```python
for encomenda in encomenda:
    try:
        cursor.execute(
            "INSERT INTO encomenda(cliente, data_pedido, data_pagamento, data_retirada, descricao) VALUES (%s, %s, %s, %s, %s);",
            (encomenda.cliente, encomenda.data_pedido, encomenda.data_pagamento, encomenda.data_retirada, encomenda.descricao)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()
```
